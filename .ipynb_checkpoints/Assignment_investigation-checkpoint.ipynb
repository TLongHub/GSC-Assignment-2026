```python
import pystac_client
import planetary_computer
import odc.stac

import rioxarray as rxr
import geopandas as gpd
import matplotlib.pyplot as plt

from shapely import box
```

```python
# Provide a name for our area of interest (AOI)
aoi_name = "okavanga"

# Provide the bounding box coordinates for the AOI in the format [min_lon, min_lat, max_lon, max_lat]
bbox = [21.6632, -20.0944, 23.9112, -18.1094]

# Provide the date range for the search in the format "YYYY-MM-DD/YYYY-MM-DD"
daterange = "2026-01-01/2026-02-28"  # rainy season 2026

# Provide a cloud cover percentage threshold for filtering the search results
max_cloud_percent = 20  # maximum cloud cover percentage

# Provide a 'collection' from the Microsoft Planetary Computer STAC API to search for imagery. 
# For example, "landsat-c2-l2" for Landsat Collection 2 Level 2 data.
# See https://planetarycomputer.microsoft.com/CATALOG for more examples
collection = "landsat-c2-l2"
```

```python
aoi = gpd.GeoDataFrame(geometry=[box(*bbox)], crs="EPSG:4326")

aoi.explore()
```

```python
# Connect to the MPC
stac_url = "https://planetarycomputer.microsoft.com/api/stac/v1"

catalog = pystac_client.Client.open(
    stac_url,
    modifier=planetary_computer.sign_inplace,  # only needed for MPC

    )
```

```python
# Search for cloud-free Landsat imagery in Summer
search = catalog.search(
    collections=[collection],
    bbox=bbox,
    datetime=daterange,
    query={
        "eo:cloud_cover": {"lt": max_cloud_percent},  # Cloud cover less than 15%
        }, 
    sortby=[{"field": "properties.eo:cloud_cover", "direction": "asc"}]
)

# Convert the search result object into a list of `pystac.Item` objects
items = search.item_collection()
print(f"Found {len(items)} matching scene(s).")
```

```python
item = items[4]
item
```

## Okavanga Delta

```python
# Requested bands - I've commented out some we don't need for now, but you can uncomment them if you want to use them later
bands = [
    # 'coastal',  # band 1, coastal/aerosol
    'blue',  # band 2, blue
    'green',  # band 3, green
    'red',  # band 4, red
    'nir08',  # band 5, near-infrared
    'swir16',  # band 6, shortwave infrared 1
    'swir22',  # band 7, shortwave infrared 2
]

# Load requested bands directly into an xarray.Dataset using odc-stac
ds = odc.stac.load(
    [item],
    bands=bands,
    bbox=bbox,
    # crs="EPSG:XXXX",  # Not used for now: by default, will use the native projection of the imagery
    # resolution=30, # Not used for now: by default, will use the native resolution of the imagery
).squeeze()

# xarray is 'lazy' by default, meaning it doesn't load the data into memory until you explicitly ask for it.
# The `.compute()` method forces the data to be loaded into memory, which can take some time depending on the size of the dataset.
ds.compute()

ds
```

```python
# stack variables into a band dimension
ds = ds.to_array(dim="band").squeeze()

# add the band names as a long_name attribute to the dataset
ds.attrs["long_name"] = list(ds["band"].values)

ds
```

```python
# set zero to NaN
ds = ds.where(ds > 0)

# Surface reflectance scaling values
ds = ds * 0.0000275 - 0.2  # convert to SR
```

```python
rgb = ds.sel(band=["red", "green", "blue"])

fig, ax = plt.subplots()
rgb.plot.imshow(ax=ax, robust=True)
ax.set_aspect("equal")
ax.set_xlim(570000, 760000)
ax.set_ylim(-2200000, -2010000)
```

```python
# Calculate NDWI
ndwi = (ds.sel(band="green") - ds.sel(band="nir08")) / (ds.sel(band="nir08") + ds.sel(band="green"))

ndwi.plot.imshow(cmap="BrBG", robust=True)
```

```python
# Calculate NDVI
ndvi = (ds.sel(band="nir08") - ds.sel(band="red")) / (ds.sel(band="nir08") + ds.sel(band="red"))

ndvi.plot.imshow(cmap="BrBG", robust=True)
```

## Functions

```python
# Requested bands - I've commented out some we don't need for now, but you can uncomment them if you want to use them later
bands = [
    # 'coastal',  # band 1, coastal/aerosol
    'blue',  # band 2, blue
    'green',  # band 3, green
    'red',  # band 4, red
    'nir08',  # band 5, near-infrared
    'swir16',  # band 6, shortwave infrared 1
    'swir22',  # band 7, shortwave infrared 2
]
```

```python
def find_sat_items(aoi_name, bbox, daterange, max_cloud_percent = 20, show = True):
    
    # Provide a 'collection' from the Microsoft Planetary Computer STAC API to search for imagery. 
    # For example, "landsat-c2-l2" for Landsat Collection 2 Level 2 data.
    # See https://planetarycomputer.microsoft.com/CATALOG for more examples
    collection = "landsat-c2-l2"

    aoi = gpd.GeoDataFrame(geometry=[box(*bbox)], crs="EPSG:4326")
    if show == True:
        aoi.explore()

    # Connect to the MPC
    stac_url = "https://planetarycomputer.microsoft.com/api/stac/v1"
    
    catalog = pystac_client.Client.open(
        stac_url,
        modifier=planetary_computer.sign_inplace,  # only needed for MPC
    
        )

    # Search for cloud-free Landsat imagery in Summer
    search = catalog.search(
        collections=[collection],
        bbox=bbox,
        datetime=daterange,
        query={
            "eo:cloud_cover": {"lt": max_cloud_percent},  # Cloud cover less than 15%
            }, 
        sortby=[{"field": "properties.eo:cloud_cover", "direction": "asc"}]
    )
    
    # Convert the search result object into a list of `pystac.Item` objects
    items = search.item_collection()
    if show == True:
        print(f"Found {len(items)} matching scene(s).")

    return items


def load_sat_data(item, bands, bbox):
    # Load requested bands directly into an xarray.Dataset using odc-stac
    ds = odc.stac.load(
        [item],
        bands=bands,
        bbox=bbox,
        # crs="EPSG:XXXX",  # Not used for now: by default, will use the native projection of the imagery
        # resolution=30, # Not used for now: by default, will use the native resolution of the imagery
    ).squeeze()
    
    # xarray is 'lazy' by default, meaning it doesn't load the data into memory until you explicitly ask for it.
    # The `.compute()` method forces the data to be loaded into memory, which can take some time depending on the size of the dataset.
    ds.compute()

    # stack variables into a band dimension
    ds = ds.to_array(dim="band").squeeze()
    
    # add the band names as a long_name attribute to the dataset
    ds.attrs["long_name"] = list(ds["band"].values)

    # set zero to NaN
    ds = ds.where(ds > 0)
    
    # Surface reflectance scaling values
    ds = ds * 0.0000275 - 0.2  # convert to SR

    return ds


def plot_sat_data(ds):
    
    rgb = ds.sel(band=["red", "green", "blue"])
    fig, ax = plt.subplots()
    rgb.plot.imshow(ax=ax, robust=True)
    ax.set_aspect("equal")


def ndbi(ds):
    ndbi = (ds.sel(band="blue") - ds.sel(band="red")) / (ds.sel(band="blue") + ds.sel(band="red"))
    ndbi.plot.imshow(cmap="BrBG", robust=True)

    return ndbi


def ndvi(ds):
    ndvi = (ds.sel(band="nir08") - ds.sel(band="red")) / (ds.sel(band="nir08") + ds.sel(band="red"))
    ndvi.plot.imshow(cmap="BrBG", robust=True)

    return ndvi

def nbr(ds):
    nbr = (ds.sel(band="nir08") - ds.sel(band="swir16")) / (ds.sel(band="nir08") + ds.sel(band="swir16"))
    nbr.plot.imshow(cmap="BrBG", robust=True)

    return nbr
```

## Okavanga Delta typical flood month

```python
bands = [
    # 'coastal',  # band 1, coastal/aerosol
    'blue',  # band 2, blue
    'green',  # band 3, green
    'red',  # band 4, red
    'nir08',  # band 5, near-infrared
    'swir16',  # band 6, shortwave infrared 1
    'swir22',  # band 7, shortwave infrared 2
]

items = find_sat_items("okavanga", [21.6632, -20.0944, 23.9112, -18.1094], "2026-08-01/2026-08-28", show = True)

ds = load_sat_data(items[4], bands)

plot_sat_data(ds)
```

```python
# Calculate NDWI
ndwi = (ds.sel(band="green") - ds.sel(band="nir08")) / (ds.sel(band="nir08") + ds.sel(band="green"))

ndwi.plot.imshow(cmap="BrBG", robust=True)
```

## Portugal wildfires

### Locating burn site

Below we find satellite imagery of the burn site on 1st July 2017. The wildfire is reported to have started on 17th June 2017 - approximately 2 weeks earlier.

```python
bbox = [-8.38, 39.78, -7.72, 40.15]
```

```python
bands = [
    # 'coastal',  # band 1, coastal/aerosol
    'blue',  # band 2, blue
    'green',  # band 3, green
    'red',  # band 4, red
    'nir08',  # band 5, near-infrared
    'swir16',  # band 6, shortwave infrared 1
    'swir22',  # band 7, shortwave infrared 2
]

bbox = [-8.38, 39.78, -7.72, 40.15]

items = find_sat_items("portugal", bbox, "2017-06-17/2017-07-15", max_cloud_percent = 20, show = True)

post_burn = load_sat_data(items[3], bands, bbox)

plot_sat_data(post_burn)
```

```python
ndvi(post_burn)
```

```python
nbr(post_burn)
```

### Looking for pre-burn evidence
